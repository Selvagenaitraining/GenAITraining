### Python Basics
#### Data Types 
* Integer
* Float
* Boolean
* Complex
* String

```python
# Integer
a=100
b=-100
print(a,type(a)) 
print(b,type(b))

# Float - 1.0 , 1e10, 123.324
a=50.55
print(a,type(a))
b2=2e7 #20000000.0 # after e represents number of zero's
print(b2,type(b2))

# Complex - 1+0j
c=1+0j #j represents complex numbers
print(c,type(c))

### String - "Hello World"

a="Hello World"
b="1000"
print(a,type(a))
print(b,type(b))

###Boolean - True , False
a=True
print(a,type(a))
```

#### Type Casting


```python
a="10"
b=10
c=int(a)
d=bool(a)
print(c+b,d,type(a),type(b),type(c))

a="hello"
b=bool(a)
print(b)

#ValueError: could not convert string to float: 'hello' 
# a1="hello"
# b1=float(a1)
# print(b1)

#Return False
e=''
f=bool(e)
print(f)

#Return True
e=' '
f=bool(e)
print(f)

#Return False
e=0
f=bool(e)
print(f)

#Return True
e=0.1
f=bool(e)
print(f)
```
