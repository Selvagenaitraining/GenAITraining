#### **Print the below pattern**

```python
# While Loop
i=1
while i<=5:
    j=1
    while j<=5:
        print("*",end='')
        j+=1
    print()
    i=i+1
```

```python
# For Loop
for i in range(1,5+1):
    for j in range(1,5+1):
        print("*",end='')
    print()
```

```python
# Generate Pattern on User Input using While Loop
a=int(input("Enter Number of Rows : "))
b=int(input("Enter Number of Columns :"))
x=input("Enter a value :")
c=1
while c<=a:
    d=1
    while d<=b:
        print(x,end="")
        d+=1
    print()
    c+=1
```

```python
# Generate Pattern on User Input using for Loop
a=int(input("Enter Number of Rows : "))
b=int(input("Enter Number of Columns :"))
x=input("Enter a value :")
for i in range(1,a+1):
    for j in range(1,b+1):
        print(x,end="")
    print()
```

```python
a = int(input("rows: "))
b = int(input("columns: "))
i = 0
while i < a:
    j = 0
    while j < b:
        print("*", end=" ")
        j += 1
    print()
    i += 1
```

```python
row=int(input("Enter the Row: "))
column=int(input("Enter the column:"))
pattern=input("Enter the pattern :")

a=1
while a<=row:
    b=1
    while b<=column:
        print(pattern,end=" ")
        b+=1
    a+=1
    print()
```
